```python
from pyspark.sql.functions import *

customers = spark.table("workspace.silver.customers_clean")
products = spark.table("workspace.silver.products_clean")
orders = spark.table("workspace.silver.orders_clean")
```

```python
print(f"Total Customers : {customers.count()}")
```

```python
customers.filter(col("customer_id").isNull()).show()
```

```python
customers.groupBy("customer_id") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```

```python
customers.filter(
    ~col("email").rlike(r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$")
).show()
```

```python
products.filter(col("price") <= 0).show()
```

```python
products.filter(col("stock_quantity") < 0).show()
```

```python
products.groupBy("product_id") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```

```python
orders.filter(col("order_id").isNull()).show()
```

```python
orders.groupBy("order_id") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```

```python
orders.filter(col("quantity") <= 0).show()
```

```python
orders.filter(col("total_amount") <= 0).show()
```

```python
invalid_customer_orders = (
    orders.join(
        customers,
        "customer_id",
        "left_anti"
    )
)

invalid_customer_orders.show()
```

```python
invalid_product_orders = (
    orders.join(
        products,
        "product_id",
        "left_anti"
    )
)

invalid_product_orders.show()
```

```python
orders.filter(
    col("total_amount") != round(col("quantity") * col("unit_price"), 2)
).show()
```
